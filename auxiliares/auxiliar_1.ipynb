**Objetivo**: La idea es usar la Ley de
Grandes Números de forma **operacional**: en vez de calcular $P(\text{ganar cambiando})$

```python
import random
import numpy as np
import matplotlib.pyplot as plt

random.seed(123)
np.random.seed(123)
```

## 1. Simulación básica ($n=3$ puertas)

Reproducimos el modelo de Monty Hall (Definición 6 de la cátedra): el premio y la elección
del jugador son uniformes e independientes en $\{1,2,3\}$; el animador conoce el premio y
siempre abre una puerta vacía distinta de la elegida (si tiene dos opciones, elige al azar
entre ellas).


```python
def jugar_cambiando():
    premio  = random.randint(1, 3)   # donde esta el premio
    elegida = random.randint(1, 3)   # puerta inicial del jugador
    # Al cambiar, el jugador gana si y solo si NO habia acertado
    return elegida != premio

N = 10000
ganadas = sum(jugar_cambiando() for _ in range(N))
print(ganadas / N)   # frecuencia estimada de ganar cambiando
```

```python
for N in [10, 100000]: # veamos qué pasa con distintos N
    ganadas = sum(jugar_cambiando() for _ in range(N))
    frecuencia = ganadas / N

    print(f"N = {N:6d}  ->  frecuencia = {frecuencia:.4f}")
```

### 2. (Propuesto) Generalización a $n$ puertas

Extendemos el modelo a $n$ puertas (Problema [insertar] de la pauta): el animador abre $n-2$
puertas vacías distintas de la elegida, dejando una sola puerta cerrada además de la
elegida. El resultado teórico derivado por Bayes es

$$P(\text{ganar cambiando}) = \frac{n-1}{n}.$$


```python
def jugar_cambiando_n(n):
    premio  = random.randint(1, n)   # donde esta el premio
    elegida = random.randint(1, n)   # puerta inicial del jugador
    # El animador abre n-2 puertas vacias; queda 1 puerta cerrada ademas
    # de la elegida. Al cambiar, uno gana si y solo si NO acerto al inicio.
    return elegida != premio
```

```python
N = 10000

for n in #num de puertas que queremos ver:
    ganadas = sum(jugar_cambiando_n(n) for _ in range(N))

    p_estimada = ganadas / N
    p_teorica = (n - 1) / n

    print(f"n = {n:2d} | estimada = {p_estimada:.4f} | teórica = {p_teorica:.4f}")
```

## 3. Convergencia: $\hat p_N$ a medida que crece $N$

La Ley de Grandes Números garantiza que $\hat p_N \to P(A)$ cuando $N\to\infty$. Graficamos
el promedio acumulado partida a partida para visualizar esta convergencia y la fluctuación
alrededor del valor teórico $2/3$.


```python
def trayectoria(N):
    """Estimaciones acumuladas p_hat_1, ..., p_hat_N de ganar cambiando (3 puertas)."""
    p_hat = []
    ganadas = 0
    for t in range(1, N + 1):
        premio  = random.randint(1, 3)
        elegida = random.randint(1, 3)
        # cambiando, se gana si y solo si NO se acerto al inicio
        if elegida != premio:
            ganadas += 1
        p_hat.append(ganadas / t)   # promedio acumulado tras t partidas
    return p_hat

N = # cuántas iteraciones?
p = trayectoria(N)
plt.plot(range(1, N + 1), p, label="estimacion")
plt.axhline(2 / 3, ls="--")   # valor teorico 2/3
```

**Qué observar:** al principio (pocas repeticiones) la curva fluctúa bastante, porque el
estimador tiene varianza alta; a medida que $N$ crece, se estabiliza cerca de $2/3$, tal como
predice la Ley de Grandes Números.

## Cierre

- La simulación confirma, sin necesidad de calcular nada analíticamente, que **cambiar de
  puerta** gana con probabilidad cercana a $2/3$, y que este valor se generaliza a
  $(n-1)/n$ con $n$ puertas.
- El error del estimador Monte Carlo decrece como $1/\sqrt N$: **cien veces más
  simulaciones solo dan diez veces más precisión**. Esto es exactamente lo que se pidió
  demostrar (y ahora verificar empíricamente) en el Problema 3 de la pauta.
- Esta misma metodología —simular, promediar, y verificar contra la Ley de Grandes
